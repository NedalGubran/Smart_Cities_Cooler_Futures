## City-Level LST Extraction Using Urban Atlas Boundaries

To obtain representative city-level land surface temperatures, the Urban Atlas Functional Urban Area (FUA) boundaries are used instead of a single city-centre pixel or a fixed buffer.

The Urban Atlas boundaries are transformed from EPSG:3035 to EPSG:4326 to match the coordinate system of the LST dataset.

LST pixels located within each city boundary are then selected and aggregated to calculate the mean daytime land surface temperature.

This approach provides a consistent spatial basis for comparing summer LST across the 10 cities from 2021 to 2024.

```python
from shapely.geometry import Point
import numpy as np

# Crop first to Berlin bounding box for speed
minx, miny, maxx, maxy = berlin_boundary.total_bounds

berlin_crop = sample.sel(
    lat=slice(miny, maxy),
    lon=slice(minx, maxx)
)

# Create lon/lat grid
lon2d, lat2d = np.meshgrid(
    berlin_crop["lon"].values,
    berlin_crop["lat"].values
)

# Berlin geometry
berlin_geom = berlin_boundary.geometry.iloc[0]

# Mask pixels inside Berlin boundary
mask = np.array([
    berlin_geom.contains(Point(x, y))
    for x, y in zip(lon2d.ravel(), lat2d.ravel())
]).reshape(lon2d.shape)

berlin_lst_city = (berlin_crop["lst"] - 273.15).where(mask)

print("Pixels inside Berlin:", int(berlin_lst_city.count()))
print("Mean LST Berlin (°C):", float(berlin_lst_city.mean(skipna=True)))
```

## Testing City-Level LST Extraction

Before processing the complete 2021–2024 dataset, the extraction method is tested for all 10 cities using June 2021.

For each city, the Urban Atlas FUA boundary is used to select LST pixels located within the city area. The mean daytime LST is then calculated in degrees Celsius.

This step validates the spatial extraction method before applying it to all summer months and years.

```python
results_test = []

for city, ua_name in ua_names.items():

    # Load Urban Atlas boundary
    ua_file = list(
        urban_atlas_dir.rglob(f"*{ua_name}*.fgb")
    )[0]

    city_ua = gpd.read_file(ua_file)

    # Create one FUA boundary and convert to LST CRS
    city_boundary = city_ua.dissolve().to_crs("EPSG:4326")
    city_geom = city_boundary.geometry.iloc[0]

    # Crop LST to city bounding box
    minx, miny, maxx, maxy = city_boundary.total_bounds

    city_crop = sample.sel(
        lat=slice(miny, maxy),
        lon=slice(minx, maxx)
    )

    # Create pixel coordinates
    lon2d, lat2d = np.meshgrid(
        city_crop["lon"].values,
        city_crop["lat"].values
    )

    # Pixels inside city boundary
    mask = np.array([
        city_geom.contains(Point(x, y))
        for x, y in zip(lon2d.ravel(), lat2d.ravel())
    ]).reshape(lon2d.shape)

    # Mean LST in Celsius
    city_lst = (city_crop["lst"] - 273.15).where(mask)

    results_test.append({
        "city": city,
        "pixels": int(city_lst.count()),
        "mean_lst_c": float(city_lst.mean(skipna=True))
    })

results_test
```

## Summer LST Extraction: 2021–2024

The validated spatial extraction method is applied to all summer LST datasets from 2021 to 2024.

For each city and month, daytime LST pixels within the Urban Atlas FUA boundary are extracted and aggregated.

The resulting dataset contains monthly city-level LST values for June, July and August across four years, providing a temporal layer for comparing summer surface heat patterns between cities.

```python
import pandas as pd
import re

# Prepare city boundaries once
city_boundaries = {}

for city, ua_name in ua_names.items():
    ua_file = list(
        urban_atlas_dir.rglob(f"*{ua_name}*.fgb")
    )[0]

    city_ua = gpd.read_file(ua_file)
    city_boundary = city_ua.dissolve().to_crs("EPSG:4326")

    city_boundaries[city] = city_boundary.geometry.iloc[0]


# Extract LST for all cities and all monthly files
lst_results = []

for file in lst_files:

    # Extract year and month from filename
    match = re.search(r"(\d{6})", file.name)
    date_code = match.group(1)

    year = int(date_code[:4])
    month = int(date_code[4:6])

    ds = xr.open_dataset(file, engine="h5netcdf")

    print(f"Processing {year}-{month:02d}...")

    for city, city_geom in city_boundaries.items():

        minx, miny, maxx, maxy = city_geom.bounds

        city_crop = ds.sel(
            lat=slice(miny, maxy),
            lon=slice(minx, maxx)
        )

        lon2d, lat2d = np.meshgrid(
            city_crop["lon"].values,
            city_crop["lat"].values
        )

        mask = np.array([
            city_geom.contains(Point(x, y))
            for x, y in zip(lon2d.ravel(), lat2d.ravel())
        ]).reshape(lon2d.shape)

        city_lst = (city_crop["lst"] - 273.15).where(mask)

        lst_results.append({
            "city": city,
            "year": year,
            "month": month,
            "pixel_count": int(city_lst.count()),
            "mean_lst_c": float(city_lst.mean(skipna=True))
        })

    ds.close()


lst_monthly = pd.DataFrame(lst_results)

print("Shape:", lst_monthly.shape)
lst_monthly.head(15)
```

```python
print("Shape:", lst_monthly.shape)
print("\nMissing values:")
print(lst_monthly.isna().sum())

print("\nPixel count range:")
print(lst_monthly["pixel_count"].describe())
```

## Annual Summer LST Summary

The monthly LST values for June, July and August are aggregated to calculate the mean summer land surface temperature for each city and year.

This creates one summer LST value per city for each year from 2021 to 2024, allowing year-to-year comparison of summer surface heat patterns.

```python
summer_lst = (
    lst_monthly
    .groupby(["city", "year"], as_index=False)
    .agg(
        summer_mean_lst_c=("mean_lst_c", "mean"),
        avg_pixel_count=("pixel_count", "mean")
    )
)

print("Shape:", summer_lst.shape)
summer_lst
```

## Summer LST Change Analysis: 2021–2024

To examine how summer surface heat changed across the study period, the annual summer mean LST values are compared between 2021 and 2024.

The analysis calculates the absolute temperature difference for each city and also identifies the warmest and coolest summer observed during the four-year period.

Because the dataset covers only four summers, the results are interpreted as recent temporal patterns rather than long-term climate trends.

```python
lst_change = (
    summer_lst
    .pivot(index="city", columns="year", values="summer_mean_lst_c")
    .reset_index()
)

lst_change["change_2021_2024_c"] = (
    lst_change[2024] - lst_change[2021]
)

lst_change["warmest_year"] = (
    lst_change[[2021, 2022, 2023, 2024]]
    .idxmax(axis=1)
)

lst_change["coolest_year"] = (
    lst_change[[2021, 2022, 2023, 2024]]
    .idxmin(axis=1)
)

lst_change = lst_change.sort_values(
    "change_2021_2024_c",
    ascending=False
)

lst_change
```

## Linking Recent Summer LST with the Urban Cooling Priority Framework

The temporal LST results are compared with the existing Priority Score and UHI intensity to examine whether cities identified as higher cooling priorities also experience higher recent summer surface temperatures.

This analysis provides an additional heat perspective without incorporating LST directly into the Priority Score.

```python
from pathlib import Path
import pandas as pd

final_path = Path("../Data_Processed/Smart_Cities_Cooler_Futures_Tableau_CLEAN.csv")

final_df = pd.read_csv(final_path)

print(final_df.shape)
print(final_df.columns.tolist())
print(final_df["city"].tolist())
```

```python
city_name_map = {
    "Athens": "ATHINA",
    "Milan": "MILANO",
    "Amsterdam": "AMSTERDAM",
    "Barcelona": "BARCELONA",
    "Madrid": "MADRID",
    "Budapest": "BUDAPEST",
    "Vienna": "WIEN",
    "Lisbon": "LISBOA",
    "Prague": "PRAHA",
    "Berlin": "BERLIN"
}

summer_lst_merge = summer_lst.copy()
summer_lst_merge["city"] = summer_lst_merge["city"].map(city_name_map)

lst_priority = summer_lst_merge.merge(
    final_df[
        ["city", "priority_score", "priority_rank", "uhi_intensity_c"]
    ],
    on="city",
    how="left"
)

print("Shape:", lst_priority.shape)

print("\nMissing after merge:")
print(
    lst_priority[
        ["priority_score", "priority_rank", "uhi_intensity_c"]
    ].isna().sum()
)

lst_priority.head(10)
```

To evaluate whether the existing prioritisation is supported by the new heat data, each city's recent summer LST is compared with its UHI intensity and Priority Score.

The four annual summer LST values are first summarized into a city-level mean. This allows the temporal LST dataset to be compared consistently with the city-level baseline indicators.

The comparison is descriptive and is used to identify whether different measures of heat and vulnerability show similar or contrasting city patterns.

```python
lst_city_summary = (
    lst_priority
    .groupby("city", as_index=False)
    .agg(
        mean_lst_2021_2024_c=("summer_mean_lst_c", "mean"),
        priority_score=("priority_score", "first"),
        priority_rank=("priority_rank", "first"),
        uhi_intensity_c=("uhi_intensity_c", "first")
    )
)

lst_city_summary = lst_city_summary.sort_values(
    "priority_rank"
)

lst_city_summary
```

## Relationship Between LST, UHI and Priority Score

Pearson correlations are calculated to explore how the new LST measure relates to UHI intensity and the existing Priority Score.

This helps determine whether cities with higher average summer surface temperatures also tend to have stronger urban heat island intensity or higher overall cooling priority.

```python
lst_uhi_corr = lst_city_summary[
    "mean_lst_2021_2024_c"
].corr(
    lst_city_summary["uhi_intensity_c"]
)

lst_priority_corr = lst_city_summary[
    "mean_lst_2021_2024_c"
].corr(
    lst_city_summary["priority_score"]
)

print(f"LST vs UHI correlation: {lst_uhi_corr:.3f}")
print(f"LST vs Priority Score correlation: {lst_priority_corr:.3f}")
```

## Summer Heat Persistence and Variability

Beyond comparing individual years, the four summer observations are used to examine how persistent and variable surface heat is across the cities.

For each city, the analysis calculates the four-year mean summer LST, the minimum and maximum summer LST, and the range between them.

This helps distinguish cities that remain consistently hot from cities that experience stronger year-to-year fluctuations.

```python
lst_patterns = (
    summer_lst
    .groupby("city", as_index=False)
    .agg(
        mean_lst_c=("summer_mean_lst_c", "mean"),
        min_lst_c=("summer_mean_lst_c", "min"),
        max_lst_c=("summer_mean_lst_c", "max"),
        lst_range_c=("summer_mean_lst_c", lambda x: x.max() - x.min())
    )
)

lst_patterns = lst_patterns.sort_values(
    "mean_lst_c",
    ascending=False
)

lst_patterns
```

## Priority Score Sensitivity Analysis

The baseline Priority Score gives equal weight to all six indicators. To test whether the city ranking depends strongly on this assumption, alternative weighting scenarios are evaluated.

Three alternative scenarios are compared with the equal-weight baseline:

- **Heat-focused:** greater emphasis on UHI intensity.
- **Exposure-focused:** greater emphasis on population density.
- **Cooling-focused:** greater emphasis on green-space and street-tree deficits.

The purpose is not to identify one “correct” weighting scheme, but to evaluate how stable the city priorities remain under different planning perspectives.

```python
score_cols = {
    "building": "building_density_per_km2_score",
    "height": "average_building_height_m_score",
    "trees": "street_tree_pct_score",
    "green": "green_space_pct_score",
    "uhi": "uhi_intensity_c_score",
    "population": "population_density_per_km2_score"
}

# Alternative weighting scenarios
weights = {
    "equal": {
        "building": 1/6, "height": 1/6, "trees": 1/6,
        "green": 1/6, "uhi": 1/6, "population": 1/6
    },
    "heat_focused": {
        "building": 0.10, "height": 0.10, "trees": 0.10,
        "green": 0.10, "uhi": 0.40, "population": 0.20
    },
    "exposure_focused": {
        "building": 0.10, "height": 0.10, "trees": 0.10,
        "green": 0.10, "uhi": 0.20, "population": 0.40
    },
    "cooling_focused": {
        "building": 0.10, "height": 0.10, "trees": 0.25,
        "green": 0.25, "uhi": 0.15, "population": 0.15
    }
}

sensitivity = final_df.copy()

for scenario, w in weights.items():
    sensitivity[f"{scenario}_score"] = sum(
        sensitivity[score_cols[k]] * weight
        for k, weight in w.items()
    )

sensitivity[
    ["city", "priority_score", "equal_score",
     "heat_focused_score", "exposure_focused_score",
     "cooling_focused_score"]
].round(3)
```

```python
scenarios = [
    "equal",
    "heat_focused",
    "exposure_focused",
    "cooling_focused"
]

for scenario in scenarios:
    sensitivity[f"{scenario}_rank"] = (
        sensitivity[f"{scenario}_score"]
        .rank(method="min", ascending=False)
        .astype(int)
    )

rank_comparison = sensitivity[
    [
        "city",
        "equal_rank",
        "heat_focused_rank",
        "exposure_focused_rank",
        "cooling_focused_rank"
    ]
].sort_values("equal_rank")

rank_comparison
```

## Priority Rank Stability

Rank stability measures how much each city's position changes across the alternative weighting scenarios.

The minimum and maximum rank are compared to calculate the total rank range. A smaller range indicates a more stable priority classification, while a larger range indicates greater sensitivity to the chosen planning perspective.

```python
rank_cols = [
    "equal_rank",
    "heat_focused_rank",
    "exposure_focused_rank",
    "cooling_focused_rank"
]

rank_stability = rank_comparison.copy()

rank_stability["best_rank"] = rank_stability[rank_cols].min(axis=1)
rank_stability["worst_rank"] = rank_stability[rank_cols].max(axis=1)

rank_stability["rank_range"] = (
    rank_stability["worst_rank"] -
    rank_stability["best_rank"]
)

rank_stability = rank_stability.sort_values(
    ["rank_range", "equal_rank"]
)

rank_stability
```

## Intervention Gap Analysis

To strengthen the link between the vulnerability assessment and recommended cooling actions, intervention gaps are calculated for green-space and street-tree coverage.

The median coverage across the 10 study cities is used as a transparent internal benchmark. For cities below the benchmark, the analysis measures how much additional green-space or street-tree coverage would be required to reach the study median.

These gaps are used as decision-support indicators rather than fixed policy targets.

```python
green_benchmark = final_df["green_space_pct"].median()
tree_benchmark = final_df["street_tree_pct"].median()

intervention_gap = final_df[
    [
        "city",
        "priority_rank",
        "green_space_pct",
        "street_tree_pct",
        "recommended_intervention"
    ]
].copy()

intervention_gap["green_space_gap_pct"] = (
    green_benchmark - intervention_gap["green_space_pct"]
).clip(lower=0)

intervention_gap["street_tree_gap_pct"] = (
    tree_benchmark - intervention_gap["street_tree_pct"]
).clip(lower=0)

print(f"Green-space benchmark: {green_benchmark:.2f}%")
print(f"Street-tree benchmark: {tree_benchmark:.2f}%")

intervention_gap.sort_values("priority_rank")
```

- **Recent summer heat patterns differ substantially across cities.** Nine of the ten cities recorded higher summer mean LST in 2024 than in 2021, while Budapest showed the greatest year-to-year variability.
- **Recent impervious-surface increase alone does not explain LST change** across the 10-city sample.
- **Some mid-ranked cities are sensitive to planning priorities.** Budapest shows the largest ranking variation across alternative weighting scenarios.
- **Intervention needs differ by city.** Green-space gaps are particularly large in Amsterdam and Milan, while street-tree gaps are more important in cities such as Madrid and Lisbon.
- The results support a **city-specific, multidimensional approach** to urban cooling rather than prioritising cities based on a single heat indicator.
